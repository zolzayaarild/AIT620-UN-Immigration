## Introduction

In this assignment, I will use Python and pandas to clean and transform the United Nations immigration dataset for Canada.

The dataset contains immigration records from 1980 to 2013.

## Objectives

1. Load the Excel dataset into a pandas DataFrame.
2. Remove unnecessary rows and columns.
3. Rename columns for better readability.
4. Clean missing values.
5. Calculate total immigration for each country.
6. Display the final DataFrame in the required format.

```python

# Step 2: Check Excel file and worksheet names

import pandas as pd
from pathlib import Path

# Find Excel files in the current folder
excel_files = list(Path(".").glob("*.xlsx"))

# Display available Excel files
print("Excel files found:")
for file in excel_files:
    print(file.name)

# Display worksheet names
if excel_files:
    excel_file = pd.ExcelFile(excel_files[0])
    print("\nAvailable worksheets:")
    print(excel_file.sheet_names)
else:
    print("No Excel file found.")
```

```python

# Step 3: Load the UN Immigration Dataset

# Read the Canada by Citizenship worksheet
# Skip the first 20 rows containing titles and headers
# Use the next row as the column headers
df = pd.read_excel(
    "Canada.xlsx",
    sheet_name="Canada by Citizenship",
    skiprows=20
)

# Display the dataset dimensions
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display the first five rows
print("\nFirst Five Rows:")
display(df.head())
```

```python

# Step 4: Remove Unnecessary Columns

# Create a copy to preserve the original dataset
df_clean = df.copy()

# Define columns that are not needed
columns_to_remove = [
    "Type",
    "Coverage",
    "AREA",
    "REG",
    "DEV"
]

# Remove unnecessary columns
df_clean = df_clean.drop(columns=columns_to_remove)

# Display the updated dataset dimensions
print("Updated Dataset Shape:", df_clean.shape)

# Display the remaining column names
print("\nRemaining Columns:")
print(df_clean.columns.tolist())

# Display the first five rows
display(df_clean.head())
```

```python

# Step 5: Rename Columns

# Rename columns to make the dataset easier to understand
df_clean = df_clean.rename(columns={
    "OdName": "Country",
    "AreaName": "Continent",
    "RegName": "Region"
})

# Display updated column names
print("Updated Column Names:")
print(df_clean.columns.tolist())

# Display the first five rows
display(df_clean.head())
```

```python

# Step 6: Set Country as the DataFrame Index

# Use country names instead of row numbers
df_clean = df_clean.set_index("Country")

# Display the index name
print("Index Name:", df_clean.index.name)

# Display the updated dataset shape
print("Dataset Shape:", df_clean.shape)

# Display the first five rows
display(df_clean.head())
```

```python

# Step 7: Calculate Total Immigration by Country

# Create a list of years from 1980 to 2013
years = list(range(1980, 2014))

# Convert immigration values to numeric format
# Replace missing or invalid values with zero
df_clean[years] = df_clean[years].apply(
    pd.to_numeric, errors="coerce"
).fillna(0)

# Calculate total immigration for each country
df_clean["Total"] = df_clean[years].sum(axis=1).astype(int)

# Display the first five countries and their totals
display(df_clean[["Continent", "Region", "DevName", "Total"]].head())

# Display the final dataset shape
print("Dataset Shape:", df_clean.shape)
```

```python

# Step 8: Validate and Display the Final DataFrame

# Define the required column order
final_columns = ["Continent", "Region", "DevName"] + years + ["Total"]

# Arrange the columns in the required order
df_final = df_clean[final_columns].copy()

# Display the final transformed dataset
print("Final Dataset Shape:", df_final.shape)
print("Index Name:", df_final.index.name)

display(df_final.head())

# Verify that the total matches the sum of yearly values
totals_match = (
    df_final[years].sum(axis=1) == df_final["Total"]
).all()

print("All immigration totals are correct:", totals_match)
```
